```python
# clase ticket, atributos,folio, problema, estado
#funciones(),leerdatos(),mostrardatos()
# creae 2 objetos ticket1, ticket2,

class TicketSoporte:
    def __init__(self):
        # Inicializamos los 3 atributos solicitados
        self.folio = ""
        self.problema = ""
        self.estado = ""

    def leerdatos(self):
        """Función para capturar datos desde el teclado."""
        print(f"\n>>> Ingresando información del Ticket")
        self.folio = input("Número de Folio: ")
        self.problema = input("Descripción del problema: ")
        self.estado = input("Estado actual (Ej: Abierto, Cerrado): ")

    def mostrardatos(self):
        """Función para imprimir los datos en pantalla."""
        print("\n==============================")
        print(f"DATOS DEL TICKET #{self.folio}")
        print("==============================")
        print(f"Problema: {self.problema}")
        print(f"Estado:   {self.estado}")
        print("------------------------------")
# --- Bloque Interactivo ---
# Creación de los 2 objetos (instancias)
ticket1 = TicketSoporte()
ticket2 = TicketSoporte()
# Interacción con el primer objeto
print("CONFIGURACIÓN TICKET 1")
ticket1.leerdatos()
# Interacción con el segundo objeto
print("\nCONFIGURACIÓN TICKET 2")
ticket2.leerdatos()
# Mostrar los resultados finales
print("\n--- RESUMEN DE SOPORTE ---")
ticket1.mostrardatos()
ticket2.mostrardatos()
print("programa Creado por Daniel Nieves NL = 41 + 3 = 44")
```
